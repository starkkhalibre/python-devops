## 4. 💡 Conceptual Cards

### 🚨 Real-World Problem
At 2 AM an API starts returning 500 errors, and your Bash `grep` script chokes on a 4 GB log file. How do you extract the failing IPs without crashing your laptop or server with out-of-memory errors? You need Python's core collections, compiled regular expressions, and lazy generators.

### Concept: Built-in Collections
**What:** Core `list`, `tuple`, `dict`, `set` types  
**Why:** Determines `O(1)` vs `O(n)` lookups and mutability  
**How:** `dict`/`set` hash keys; `tuple` is immutable  
**When:** Use `dict` for configs, `set` for deduplication

📖 *Book listing:* Sequence & Dict Operations (pp. 49–58)

**Plain-language analogy:** Think of a `list` like a numbered checklist, a `dict` like a phonebook (look up a name instantly without reading every page), and a `set` like a guest list that automatically rejects duplicates.

**Walkthrough:** The book creates lists of hosts, deduplicates IPs with `set()`, and looks up optional service ports with `dict.get()`. Using `.get(key, default)` avoids `KeyError` crashes when a config key is absent.

### Concept: Compiled Regex & Generators
**What:** `re.compile()` + `yield` functions  
**Why:** Prevents OOM on multi-GB log files  
**How:** Compiles pattern once; yields one item at a time  
**When:** Any log/stream larger than a few MBs

📖 *Book listing:* Regular Expressions & Lazy Generators (pp. 69–77)

**Plain-language analogy:** A compiled regex is like a custom stencil cut once and reused on thousands of pages; a `yield` generator is like reading a book page-by-page instead of photocopying the whole library onto your desk.

**Walkthrough:** The book introduces `re.search` and compiled regexes with named groups `(?P<name>...)` (pp. 69–73) and `yield` generators for lazy evaluation (pp. 76–77), which we apply here to parse Apache access logs in `O(1)` memory.

#### Example 5: Intermediate Example 2: Compiled regex log parser + lazy generator
*Expected output:* Printed labelled output below.

```python
# Intermediate Example 2: Compiled regex log parser + lazy generator
import re

LOG_PATTERN = re.compile(
    r'(?P<ip>\d+\.\d+\.\d+\.\d+) - (?P<user>\w+) '
    r'\[(?P<time>[^\]]+)\] "(?P<method>\w+) (?P<path>\S+) [^"]+" (?P<status>\d{3})'
)

def parse_logs(lines: list[str]):
    for line in lines:
        if match := LOG_PATTERN.search(line):
            yield match.groupdict()

sample_logs = [
    '10.0.0.5 - alice [27/Nov/2026:10:00:01] "GET /api/health HTTP/1.1" 200',
    '10.0.0.9 - bob [27/Nov/2026:10:00:02] "POST /api/deploy HTTP/1.1" 500',
]
for entry in parse_logs(sample_logs):
    print('Parsed:', entry)
```

## 8. 🚀 Real-World DevOps Applications (Personal DevOps Learning Track)
*Supplementary extension:* **Log-Redaction Pipeline**—scrubbing IPv4 addresses and secret tokens before shipping logs to centralized storage.

#### Example 6: import re
*Expected output:* Printed labelled output below.

```python
import re

IP_RE = re.compile(r'\b\d{1,3}(?:\.\d{1,3}){3}\b')
TOKEN_RE = re.compile(r'(?i)(token|apikey|password)=\S+')

def sanitize_log_stream(lines: list[str]):
    for line in lines:
        redacted = IP_RE.sub('<REDACTED_IP>', line)
        yield TOKEN_RE.sub(r'\1=<REDACTED>', redacted)

for clean in sanitize_log_stream(['INFO login from 192.168.1.42 token=secret123']):
    print(clean)
```

## 9. ⚠️ Common Mistakes & Troubleshooting

- ❌ **Bare `except:`** -> Swallows `KeyboardInterrupt` & hides bugs. ✅ Catch specific exceptions (`IndexError`, `ValueError`).
- ❌ **Iterating a generator twice** -> Second `list(gen)` returns `[]`. ✅ Materialize once or re-create the generator.
- ❌ **Missing `r'...'` in regex** -> `\d` or `\n` misinterpreted. ✅ Always prefix regex patterns with `r''`.

**Q1:** What is the difference between `/` and `//` in Python?

<details>
<summary>✅ Answer 1</summary>

`/` is true (float) division; `//` is integer floor division.

</details>

**Q2:** Why should you prefer `my_dict.get('key', default)` over `my_dict['key']`?

<details>
<summary>✅ Answer 2</summary>

`.get()` returns a safe default instead of raising `KeyError` when a key is absent.

</details>

**Q3:** Why are generators (`yield`) preferred over list comprehensions for 10 GB logs?

<details>
<summary>✅ Answer 3</summary>

Generators yield one item at a time (`O(1)` memory) instead of loading all lines into RAM.

</details>

**Q4:** What does `re.compile()` do, and why use raw strings (`r'...'`)?

<details>
<summary>✅ Answer 4</summary>

`re.compile()` pre-compiles patterns for faster reuse; `r'...'` stops Python from escaping backslashes.

</details>

**Q5:** Why is bare `except:` an anti-pattern?

<details>
<summary>✅ Answer 5</summary>

It catches `KeyboardInterrupt`/`SystemExit` and hides unexpected bugs.

</details>

```python
# Complete Solutions + Assert Verification
import re

def format_server_banner(hostname: str, env: str = 'prod') -> str:
    return f'[{env.strip().upper()}] {hostname.strip().upper()}'

def classify_case(text: str) -> str:
    if not text:
        return 'empty'
    if text.isupper():
        return 'uppercase'
    if text.islower():
        return 'lowercase'
    return 'mixed'

def capitalize_letters(word: str) -> list[str]:
    return [ch.upper() for ch in word]

def even_odd():
    while True:
        yield 'Even'
        yield 'Odd'

def parse_ports(raw_ports: list[str]) -> list[int]:
    valid: list[int] = []
    for raw in raw_ports:
        try:
            port = int(raw)
            if 1 <= port <= 65535:
                valid.append(port)
        except ValueError:
            continue
    return valid

ACCESS_RE = re.compile(r'(?P<ip>\d+\.\d+\.\d+\.\d+).*?" (?P<status>\d{3})\b')

def analyze_access_logs(lines: list[str]) -> dict:
    total, errors_5xx, ips = 0, 0, set()
    for line in lines:
        if m := ACCESS_RE.search(line):
            total += 1
            ips.add(m.group('ip'))
            if m.group('status').startswith('5'):
                errors_5xx += 1
    rate = round((errors_5xx / total) * 100, 2) if total else 0.0
    return {'total': total, 'errors_5xx': errors_5xx, 'error_rate_pct': rate, 'unique_ips': sorted(ips)}

assert format_server_banner('  web-01 ') == '[PROD] WEB-01'
assert classify_case('HELLO') == 'uppercase' and classify_case('hi') == 'lowercase' and classify_case('') == 'empty'
assert capitalize_letters('smogtether') == ['S', 'M', 'O', 'G', 'T', 'E', 'T', 'H', 'E', 'R']
eo = even_odd()
assert [next(eo) for _ in range(4)] == ['Even', 'Odd', 'Even', 'Odd']
assert parse_ports(['22', 'bad', '443', '99999']) == [22, 443]
assert analyze_access_logs(['10.0.0.1 - - "GET /" 200', '10.0.0.2 - - "POST /" 502'])['error_rate_pct'] == 50.0
print('All Chapter 1 assertions PASSED!')
```
